# Encontrar inconsistências nas bases

Verficar os valores recebidos pela seguradora e cruzar os dados com os valores do nosso banco pra encontrar inconsistências

```python
import pandas as pd

# importa os valores do extrato da seguradora
extrato_seguradora = pd.read_csv('extrato-seguradora.csv', dtype={'cpf': str})
# display(extrato_seguradora.head())

# separa o extrato por empresa
extrato_vertex = extrato_seguradora[extrato_seguradora['empresa'] == 'Vertex Logística']
extrato_sopro = extrato_seguradora[extrato_seguradora['empresa'] == 'Sopro Alimentos']
extrato_aurora = extrato_seguradora[extrato_seguradora['empresa'] == 'Aurora Clínicas']

# retira a coluna empresa e o indice do dataframe
extrato_vertex = extrato_vertex.reset_index(drop=True)
extrato_sopro = extrato_sopro.reset_index(drop=True)
extrato_aurora = extrato_aurora.reset_index(drop=True)

display(extrato_vertex.head())
display(extrato_sopro.head())
display(extrato_aurora.head())
```

```python
# importa os valores das contribuições de agosto de cada empresa
contribuicoes_vertex = pd.read_excel('contribuicoes-agosto.xlsx', sheet_name='Vertex Logística', dtype={'cpf': str})
contribuicoes_sopro = pd.read_excel('contribuicoes-agosto.xlsx', sheet_name='Sopro Alimentos', dtype={'cpf': str})
contribuicoes_aurora = pd.read_excel('contribuicoes-agosto.xlsx', sheet_name='Aurora Clínicas', dtype={'cpf': str})

display(contribuicoes_vertex.head())
# display(contribuicoes_sopro.head())
# display(contribuicoes_aurora.head())
```

```python
# df_colaboradores: cpf e salario

# importa os dados dos colaboradores que estão na folha de pagamento
colaboradores_vertex = pd.read_csv('colaboradores_vertex.csv', dtype={'cpf': str})
colaboradores_sopro = pd.read_csv('colaboradores_sopro.csv', dtype={'cpf': str})
colaboradores_aurora = pd.read_csv('colaboradores_aurora.csv', dtype={'cpf': str})

# remove as colunas nome, empresa, data_admissao, data_desligamento, status e o indice do dataframe
colaboradores_vertex = colaboradores_vertex.drop(columns=['nome', 'empresa', 'data_admissao', 'data_desligamento', 'status']).reset_index(drop=True)
colaboradores_sopro = colaboradores_sopro.drop(columns=['nome', 'empresa', 'data_admissao', 'data_desligamento', 'status']).reset_index(drop=True)
colaboradores_aurora = colaboradores_aurora.drop(columns=['nome', 'empresa', 'data_admissao', 'data_desligamento', 'status']).reset_index(drop=True)


display(colaboradores_vertex.head())
# display(colaboradores_sopro.head())
# display(colaboradores_aurora.head())
```

```python
# cria um DataFrame consolidado com as contribuições e contrapartidas de cada empresa
df_vertex = colaboradores_vertex.merge(contribuicoes_vertex, on='cpf', how='left')
df_sopro = colaboradores_sopro.merge(contribuicoes_sopro, on='cpf', how='left')
df_aurora = colaboradores_aurora.merge(contribuicoes_aurora, on='cpf', how='left')

display(df_vertex.head())
# display(df_sopro.head())
# display(df_aurora.head())
```

```python
# crio um dataframe juntando o df da empresa com o extrato da seguradora, para comparar os valores de contribuições e contrapartidas com os valores do extrato
comparativo_vertex = df_vertex.merge(extrato_vertex, on='cpf', how='left', suffixes=('_agosto', '_extrato'))
comparativo_sopro = df_sopro.merge(extrato_sopro, on='cpf', how='left', suffixes=('_agosto', '_extrato'))
comparativo_aurora = df_aurora.merge(extrato_aurora, on='cpf', how='left', suffixes=('_agosto', '_extrato'))

display(comparativo_vertex.head())
# display(comparativo_sopro.head())
# display(comparativo_aurora.head())
```

```python
# Filtra apenas as linhas onde os valores são diferentes
comparativo_vertex['diferenca_contribuicao'] = (
    comparativo_vertex['contribuicao_colaborador_agosto'] - comparativo_vertex['contribuicao_colaborador_extrato']
)

comparativo_aurora['diferenca_contribuicao'] = (
    comparativo_aurora['contribuicao_colaborador_agosto'] - comparativo_aurora['contribuicao_colaborador_extrato']
)

comparativo_sopro['diferenca_contribuicao'] = (
    comparativo_sopro['contribuicao_colaborador_agosto'] - comparativo_sopro['contribuicao_colaborador_extrato']
)

# Filtra apenas as linhas com diferença
diferencas_vertex = comparativo_vertex[comparativo_vertex['diferenca_contribuicao'] != 0]
diferencas_aurora = comparativo_aurora[comparativo_aurora['diferenca_contribuicao'] != 0]
diferencas_sopro = comparativo_sopro[comparativo_sopro['diferenca_contribuicao'] != 0]


print(diferencas_vertex.to_string(index=False))
print(diferencas_aurora.to_string(index=False))
print(diferencas_sopro.to_string(index=False))
```

```python
# Filtra linhas onde a contribuição do extrato está ausente (NaN)
nao_encontrados_vertex = comparativo_vertex[
    comparativo_vertex['contribuicao_colaborador_extrato'].isna()
]

nao_encontrados_sopro = comparativo_sopro[
    comparativo_sopro['contribuicao_colaborador_extrato'].isna()
]

nao_encontrados_aurora = comparativo_aurora[
    comparativo_aurora['contribuicao_colaborador_extrato'].isna()
]

print(nao_encontrados_vertex.to_string(index=False))
print(nao_encontrados_sopro.to_string(index=False))
print(nao_encontrados_aurora.to_string(index=False))
```

```python
# exporta o comparativo para um arquivo csv

comparativo_vertex.to_csv('comparativo_vertex.csv', index=False)
comparativo_sopro.to_csv('comparativo_sopro.csv', index=False)
comparativo_aurora.to_csv('comparativo_aurora.csv', index=False)
```
